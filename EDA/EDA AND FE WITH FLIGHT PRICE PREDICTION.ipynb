# EDA AND FEATURE ENGINEERING FLIGHT PRICE PREDICTION

In simple terms , EDA is a process to kiik at your data carefully before doing any serious work with it. You check what the data looks like, how big it is, if anything is missing or strange, what the average numbers are, how the values are spread out, and whether some things seem to move together — usually by looking at the first few rows, making simple tables of summary numbers, and drawing easy charts like bars or histograms.

The whole point is to understand the data, spot problems early, and get ideas about what might be interesting or important — before you start building models or making big conclusions. 

Data Set Information:

### FEATURES
The various features of the cleaned dataset are explained below:

1) Airline: The name of the airline company is stored in the airline column. It is a categorical feature having 6 different airlines.
2) Flight: Flight stores information regarding the plane's flight code. It is a categorical feature.
3) Source City: City from which the flight takes off. It is a categorical feature having 6 unique cities.
4) Departure Time: This is a derived categorical feature obtained created by grouping time periods into bins. It stores information about the departure time and have 6 unique time labels.
5) Stops: A categorical feature with 3 distinct values that stores the number of stops between the source and destination cities.
6) Arrival Time: This is a derived categorical feature created by grouping time intervals into bins. It has six distinct time labels and keeps information about the arrival time.
7) Destination City: City where the flight will land. It is a categorical feature having 6 unique cities.
8) Class: A categorical feature that contains information on seat class; it has two distinct values: Business and Economy.
9) Duration: A continuous feature that displays the overall amount of time it takes to travel between cities in hours.
10)Days Left: This is a derived characteristic that is calculated by subtracting the trip date by the booking date.
11) Price: Target variable stores information of the ticket price.

```python
# How to handel the data column
"""
So do so , we need the help of feature enginerring .
With the help of this we can seperate the every part 
of the data i.e date , month, year.
"""

## Feature engineering
df['Date'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
"""
Now the problem is that , after the slpiting also 
the data type of data is still not change, so what
we have to do is that , we have to typecast the data type 
into int .
"""

df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
"""
Now since we have converted the column
'Date_of_Journey', we don't need this column
anymore , so we will drop this .
"""

df.drop('Date_of_Journey', axis=1,inplace=True)
```

```python
df.head()
```

Before going to execuit the same process we have to apply 
lambda function to exract the second part of the arrival time.
This is because if we apply direct the split part then the second
part will also be in the column i.e like 22 may, 18 march etc. 
So avoid that first we have to apply the lamda function so that 
we can get only hour and minute.

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ' )[0])
```

```python
"""
Now next this we have to see that we have arrivarl and departure 
time well, so we need to extract that data as well.
"""

df['Arrival_hour'] = df['Arrival_Time'].str.split(':').str[0]
df['Arrival_min'] = df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df['Arrival_hour'] = df['Arrival_hour'].astype(int)
df['Arrival_min'] = df['Arrival_min'].astype(int)
```

```python
df.drop('Arrival_Time',axis =1, inplace= True)
```

```python
df.head(2)
```

```python
"""
Now same you have to do with departure time aswell and with duration time as well
"""

df['Departure_hour'] = df['Dep_Time'].str.split(':').str[0]
df['Departure_min'] = df['Dep_Time'].str.split(':').str[1]
df.info()
```

```python
df['Departure_hour']= df['Departure_hour'].astype(int)
df['Departure_min']= df['Departure_min'].astype(int)
df.info()
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

So what we will do here is that we will assign some labels to the stop values.
But before we have to find how many NAN values are there.

```python
"""
So what we will do is that , we will using some map function to 
replace the NAN value with 1 stop . And it goes the same for other 
as well. We will convert the 'non stop' into 0,'1 stop' into 1 
and the others well.

We are doing this so that we can convert the categorical feature 
into numerical type as well as my machine model will be able to
understand what is relation between each other .
"""


df['Total_Stops']= df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stop':2,'3 stop':3,
                                          '4 stop':4, np.nan:1})
```
